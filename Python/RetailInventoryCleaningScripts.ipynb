```python
import pandas as pd
import numpy as np

#Loaded a data
inventory = pd.read_csv("Inventory_Transactions.csv")
products = pd.read_csv("Product_Details.csv")
locations = pd.read_csv("Inventory_Locations.csv")
print("Data Loaded")
```

```python
#Checked Shapes
print("\nInventory Shape:")
print(inventory.shape)

print("\nProduct Shape:")
print(products.shape)

print("\nLocation Shape:")
print(locations.shape)
```

```python
#checked Columns
print("\nInventory Columns:")
print(inventory.columns)

print("\nProduct Columns:")
print(products.columns)

print("\nLocation Columns:")
print(locations.columns)
```

```python
print("\nInventory Data Types:")
print(inventory.dtypes)

print("\nProduct Data Types:")
print(products.dtypes)

print("\nLocation Data Types:")
print(locations.dtypes)
```

```python
#cleaned column names
inventory.columns = (
    inventory.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

products.columns = (
    products.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

locations.columns = (
    locations.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)
print("Cleaned Column Names")
```

```python
print("\nMissing Values - Inventory")
print(inventory.isnull().sum())

print("\nMissing Values - Products")
print(products.isnull().sum())

print("\nMissing Values - Locations")
print(locations.isnull().sum())
```

```python
print("\nDuplicate rows in Inventory:")
print(inventory.duplicated().sum())

print("\nDuplicate rows in Products:")
print(products.duplicated().sum())

print("\nDuplicate rows in Locations:")
print(locations.duplicated().sum())
```

```python
inventory["inventory_id"] = (
    inventory["inventory_id"]
    .astype(str)
    .str.strip()
)

inventory["product_id"] = (
    inventory["product_id"]
    .astype(str)
    .str.strip()
)

inventory["stock_out"] = (
    inventory["stock_out"]
    .astype(str)
    .str.strip()
)

inventory["overstock"] = (
    inventory["overstock"]
    .astype(str)
    .str.strip()
)
print("Cleaned Inventory Text Columns")
```

```python
text_columns_locations = [
    "inventory_id",
    "location_id",
    "location_name",
    "location_type",
    "city",
    "channel"
]

for column in text_columns_locations:
    locations[column] = locations[column].astype(str).str.strip()


text_columns_products = [
    "product_id",
    "product_name",
    "category",
    "subcategory",
    "size",
    "color",
    "season",
    "supplier",
    "product_status"
]

for column in text_columns_products:
    products[column] = products[column].astype(str).str.strip()
print("Cleaned Product and Location Text Columns")
```

```python
inventory["stock_out"] = (
    inventory["stock_out"]
    .replace({
        "Y": "Yes",
        "N": "No",
        "YES": "Yes",
        "NO": "No",
        "yes": "Yes",
        "no": "No"
    })
)

inventory["overstock"] = (
    inventory["overstock"]
    .replace({
        "Y": "Yes",
        "N": "No",
        "YES": "Yes",
        "NO": "No",
        "yes": "Yes",
        "no": "No"
    })
)
print("Standrized Yes and No Values")
```

```python
inventory["date"] = pd.to_datetime(
    inventory["date"],
    format="mixed",
    dayfirst=True,
    errors="coerce"
)

print("\nInvalid / Missing Dates:")
print(inventory["date"].isnull().sum())
```

```python
numeric_columns = [
    "opening_stock",
    "stock_recieved",
    "stock_transfer_in",
    "stock_transfer_out",
    "unit_sold",
    "unit_returned",
    "damaged_stock",
    "unit_cost",
    "sales_amount"
]

for column in numeric_columns:
    inventory[column] = pd.to_numeric(
        inventory[column],
        errors="coerce"
    )

print("Checked Numeric Column")
```

```python
#checking missing values after conversion
print("\nMissing values in transaction numeric columns:")

print(
    inventory[numeric_columns].isnull().sum()
)
```

```python
#checking negative values
print("\nNegative Opening Stock:")
print(
    inventory[
        inventory["opening_stock"] < 0
    ].shape[0]
)

print("\nNegative Units Sold:")
print(
    inventory[
        inventory["unit_sold"] < 0
    ].shape[0]
)

print("\nNegative Sales Amount:")
print(
    inventory[
        inventory["sales_amount"] < 0
    ].shape[0]
)
```

```python
#check product price
print("\nProducts with invalid price:")

print(
    products[
        products["unit_price"] <= 0
    ]
)
```

```python
print("\nProduct Categories:")
print(products["category"].unique())

print("\nSubcategories:")
print(products["subcategory"].unique())

print("\nProduct Status:")
print(products["product_status"].unique())

print("\nInventory Stock Out Values:")
print(inventory["stock_out"].unique())

print("\nInventory Overstock Values:")
print(inventory["overstock"].unique())
```

```python
duplicate_count = inventory.duplicated().sum()

print("\nDuplicate inventory transaction rows:", duplicate_count)

inventory = inventory.drop_duplicates()
```

```python
inventory.insert(
    0,
    "transaction_id",
    ["TXN" + str(i).zfill(6)
     for i in range(1, len(inventory) + 1)]
)
```

```python
inventory["closing_stock"] = (
    inventory["opening_stock"]
    + inventory["stock_recieved"]
    + inventory["stock_transfer_in"]
    - inventory["stock_transfer_out"]
    - inventory["unit_sold"]
    - inventory["damaged_stock"]
    + inventory["unit_returned"]
)

print("Created closing stock column")
```

```python
negative_closing_stock = inventory[
    inventory["closing_stock"] < 0
]

print("\nNegative Closing Stock Records:")
print(len(negative_closing_stock))

print(
    negative_closing_stock[
        [
            "inventory_id",
            "product_id",
            "opening_stock",
            "stock_recieved",
            "stock_transfer_in",
            "stock_transfer_out",
            "unit_sold",
            "damaged_stock",
            "unit_returned",
            "closing_stock"
        ]
    ].head()
)
```

```python
inventory["stock_issue"] = np.where(
    inventory["closing_stock"] < 0,
    "Negative Stock",
    "OK"
)
inventory["stock_issue"].value_counts()
```

```python
#Displaying Negetive closing stock rows
negative_stock = inventory[
    inventory["closing_stock"] < 0
]

negative_stock[
    ["inventory_id", "product_id", "date",
     "opening_stock", "stock_recieved",
     "stock_transfer_in", "stock_transfer_out",
     "unit_sold", "unit_returned",
     "damaged_stock", "closing_stock"]
]
```

```python
print("\n====================================")
print("FINAL DATASET INFORMATION")
print("====================================")

print("\nInventory:")
print(inventory.shape)

print("\nProducts:")
print(products.shape)

print("\nLocations:")
print(locations.shape)

print("\nMissing Values - Inventory:")
print(inventory.isnull().sum())

print("\nCleaning Completed Successfully!")
```

```python
#fact checking
invalid_product_ids = inventory[
    ~inventory["product_id"].isin(
        products["product_id"]
    )
]

print("\nInvalid Product IDs in Transactions:")
print(len(invalid_product_ids))

invalid_inventory_ids = inventory[
    ~inventory["inventory_id"].isin(
        locations["inventory_id"]
    )
]

print("\nInvalid Inventory IDs in Transactions:")
print(len(invalid_inventory_ids))
```

```python
inventory.to_csv(
    "Inventory_Transactions_Cleaned.csv",
    index=False
)

products.to_csv(
    "Product_Details_Cleaned.csv",
    index=False
)

locations.to_csv(
    "Inventory_Locations_Cleaned.csv",
    index=False
)

print("File saved successfully")
```
